```python
# Q1) Create a dictionary + Pandas DataFrame

import pandas as pd

# dataset
Data = {
    "Heart_Rate"  : [70, 120, 80, 130, 72, 110],
    "Temperature" : [37.0, 39.5, 36.8, 40.0, 37.2, 38.9],
    "Status"      : [0, 1, 0, 1, 0, 1]
}

df = pd.DataFrame(Data)
print(df)
```

```python
# Q2) Build Sequential Model using TensorFlow
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorfloe.keras.layers import Denes

model = sequential([
    Denes(8 , activation = 'relu' , input_shape=(2,)),
    Denes(1 , activation = 'sigmoid')
])
```

```python
# Q3) Compile using Adam + Binary Crossentropy
model.compile(
    optimizer = 'adam',
    loss = 'binary_crossentropy',
    metrics = ['accuracy']
)
```

```python
# Q4) Train for 100 epochs + plot
x = df[["Heart_Rate", "Temperature"]]
y = df["Status"]
history = model.fit(
    x , y,
    epochs = 100,
    verbos = 0
)

import matplotlib.pyplot as plt

plt.plot(history.history['loss'])
plt.title("Model Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()

plt.plot(history.history['accuracy'])
plt.title("Model Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.show()
```

```python
# Q5) Predict first patient
first_patient = [[70, 37.0]]
prediction = model.predict(first_patient)
print(prediction)

if prediction[0][0] > 0.5:
    print("Emergency")
else:
    print("Stable")
```
